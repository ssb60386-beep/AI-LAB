```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris

# ------------------------------------------------
# 1. Load Dataset
# ------------------------------------------------

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df['species'] = iris.target

# Convert numerical labels to class names
df['species'] = df['species'].map({
    0: 'setosa',
    1: 'versicolor',
    2: 'virginica'
})

# ------------------------------------------------
# 2. Dataset Information
# ------------------------------------------------

print("First five records:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nBasic Statistics:")
print(df.describe())

# ------------------------------------------------
# 3. NumPy Statistical Analysis
# ------------------------------------------------

features = df.iloc[:, 0:4].values

feature_names = [
    'Sepal Length',
    'Sepal Width',
    'Petal Length',
    'Petal Width'
]

print("\nNumPy Statistical Analysis")

for i in range(4):

    print("\n", feature_names[i])

    print("Mean:",
          np.mean(features[:, i]))

    print("Minimum:",
          np.min(features[:, i]))

    print("Maximum:",
          np.max(features[:, i]))

    print("Standard Deviation:",
          np.std(features[:, i]))

# ------------------------------------------------
# 4. Pandas Data Manipulation
# ------------------------------------------------

print("\nSelected Columns:")
print(df[
    ['petal length (cm)',
     'petal width (cm)',
     'species']
].head())

print("\nFiltered Records:")
print(
    df[df['petal length (cm)'] > 5]
)

print("\nClass Counts:")
print(df['species'].value_counts())

print("\nGrouped Mean:")
print(
    df.groupby('species').mean(numeric_only=True)
)

# ------------------------------------------------
# 5. Summary Statistics by Class
# ------------------------------------------------

summary = df.groupby('species').agg({
    'sepal length (cm)': ['mean', 'min', 'max'],
    'sepal width (cm)': ['mean', 'min', 'max'],
    'petal length (cm)': ['mean', 'min', 'max'],
    'petal width (cm)': ['mean', 'min', 'max']
})

print("\nSummary Statistics:")
print(summary)

# ------------------------------------------------
# 6. Visualization 1
# Petal Length vs Petal Width
# ------------------------------------------------

plt.figure(figsize=(8, 6))

for species in df['species'].unique():

    data = df[df['species'] == species]

    plt.scatter(
        data['petal length (cm)'],
        data['petal width (cm)'],
        label=species
    )

plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.title('Petal Length vs Petal Width')
plt.legend()
plt.grid(True)
plt.show()

# ------------------------------------------------
# 7. Visualization 2
# Sepal Length vs Sepal Width
# ------------------------------------------------

plt.figure(figsize=(8, 6))

for species in df['species'].unique():

    data = df[df['species'] == species]

    plt.scatter(
        data['sepal length (cm)'],
        data['sepal width (cm)'],
        label=species
    )

plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.title('Sepal Length vs Sepal Width')
plt.legend()
plt.grid(True)
plt.show()

# ------------------------------------------------
# 8. Visualization 3
# Petal Length Box Plot
# ------------------------------------------------

plt.figure(figsize=(8, 6))

df.boxplot(
    column='petal length (cm)',
    by='species'
)

plt.title('Petal Length Distribution by Species')
plt.suptitle('')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.show()

# ------------------------------------------------
# 9. Correlation Analysis
# ------------------------------------------------

correlation = df.iloc[:, 0:4].corr()

print("\nCorrelation Matrix:")
print(correlation)

plt.figure(figsize=(8, 6))

plt.imshow(correlation, cmap='viridis')

plt.colorbar()

plt.xticks(
    range(4),
    feature_names,
    rotation=45
)

plt.yticks(
    range(4),
    feature_names
)

plt.title('Feature Correlation Matrix')

plt.show()

# ------------------------------------------------
# 10. Prepare Data for Machine Learning
# ------------------------------------------------

X = df.iloc[:, 0:4]
y = df['species']

print("\nMachine Learning Features:")
print(X.head())

print("\nMachine Learning Target:")
print(y.head())
```

```python
# ============================================
# QUESTION 2 - MACHINE LEARNING USING IRIS
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score


# --------------------------------------------
# 1. Load Iris Dataset
# --------------------------------------------

iris = load_iris()

X = iris.data
y = iris.target

df = pd.DataFrame(
    X,
    columns=iris.feature_names
)

df["species"] = iris.target_names[y]

print("First five rows:")
print(df.head())


# --------------------------------------------
# 2. Identify Input Features
# --------------------------------------------

print("\nInput Features:")
for feature in iris.feature_names:
    print(feature)


# --------------------------------------------
# 3. Identify Target Variable
# --------------------------------------------

print("\nTarget Variable: Iris Species")
print("Classes:", iris.target_names)


# --------------------------------------------
# 4. Supervised Learning Visualization
# --------------------------------------------

plt.figure(figsize=(8, 6))

for species in iris.target_names:
    index = iris.target_names[y] == species

    plt.scatter(
        X[index, 2],
        X[index, 3],
        label=species
    )

plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("Iris Classification Features")
plt.legend()
plt.show()


# --------------------------------------------
# 5. Prepare Data for Unsupervised Learning
# --------------------------------------------

X_unsupervised = df[
    [
        "sepal length (cm)",
        "sepal width (cm)",
        "petal length (cm)",
        "petal width (cm)"
    ]
]


# --------------------------------------------
# 6. Standardize Features
# --------------------------------------------

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_unsupervised)


# --------------------------------------------
# 7. K-Means Clustering
# --------------------------------------------

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)

df["Cluster"] = clusters


# --------------------------------------------
# 8. Display Cluster Results
# --------------------------------------------

print("\nCluster Results:")
print(df.head())


# --------------------------------------------
# 9. Silhouette Score
# --------------------------------------------

score = silhouette_score(
    X_scaled,
    clusters
)

print("\nSilhouette Score:", score)


# --------------------------------------------
# 10. Visualize Unsupervised Clusters
# --------------------------------------------

plt.figure(figsize=(8, 6))

plt.scatter(
    X_unsupervised["petal length (cm)"],
    X_unsupervised["petal width (cm)"],
    c=clusters
)

plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("K-Means Unsupervised Clustering")
plt.show()


# --------------------------------------------
# 11. Actual Classes for Comparison
# --------------------------------------------

plt.figure(figsize=(8, 6))

for species in iris.target_names:
    index = iris.target_names[y] == species

    plt.scatter(
        X[index, 2],
        X[index, 3],
        label=species
    )

plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("Actual Iris Classes")
plt.legend()
plt.show()
```
