```python
def crop_decision(N, P, K, temperature, humidity, ph, rainfall):

    score = 0

    # Nitrogen condition
    if N >= 40:
        score += 1
    elif N >= 20:
        score += 0.5
    else:
        score += 0

    # Phosphorus condition
    if P >= 30:
        score += 1
    elif P >= 15:
        score += 0.5

    # Potassium condition
    if K >= 30:
        score += 1
    elif K >= 15:
        score += 0.5

    # Temperature condition
    if 20 <= temperature <= 30:
        score += 1
    elif 15 <= temperature <= 35:
        score += 0.5

    # Humidity condition
    if 50 <= humidity <= 80:
        score += 1
    elif 40 <= humidity <= 90:
        score += 0.5

    # pH condition
    if 5.5 <= ph <= 7.5:
        score += 1
    elif 5.0 <= ph <= 8.0:
        score += 0.5

    # Rainfall condition
    if 80 <= rainfall <= 250:
        score += 1
    elif 50 <= rainfall <= 300:
        score += 0.5

    # Final decision
    if score >= 5:
        decision = "Highly Suitable"
    elif score >= 3:
        decision = "Moderately Suitable"
    else:
        decision = "Less Suitable"

    return decision
```

```python
decision = crop_decision(
    90,      # N
    42,      # P
    43,      # K
    25.5,    # Temperature
    70,      # Humidity
    6.5,     # pH
    180      # Rainfall
)

print("Agricultural Decision:", decision)
```

```python
decisions = []

for index, row in df.iterrows():

    decision = crop_decision(
        row["N"],
        row["P"],
        row["K"],
        row["temperature"],
        row["humidity"],
        row["ph"],
        row["rainfall"]
    )

    decisions.append(decision)

df["rule_decision"] = decisions

print("Rule-based decisions generated successfully!")
```

```python
def recommend_crop(N, P, K, temperature, humidity, ph, rainfall):

    # Rice-like conditions
    if (
        humidity >= 70
        and rainfall >= 150
        and temperature >= 20
        and ph >= 5.0
        and ph <= 7.5
    ):
        return "rice"

    # Maize-like conditions
    elif (
        temperature >= 20
        and temperature <= 30
        and rainfall >= 60
        and rainfall <= 150
        and humidity >= 50
    ):
        return "maize"

    # Chickpea-like conditions
    elif (
        temperature >= 15
        and temperature <= 30
        and rainfall < 100
        and ph >= 6
        and ph <= 8
    ):
        return "chickpea"

    # Cotton-like conditions
    elif (
        temperature >= 25
        and humidity >= 50
        and rainfall >= 50
        and rainfall <= 150
    ):
        return "cotton"

    # Generic vegetable category
    elif (
        temperature >= 18
        and temperature <= 30
        and ph >= 5.5
        and ph <= 7.5
    ):
        return "vegetable"

    else:
        return "general crop"
```

```python
crop_predictions = []

for index, row in df.iterrows():

    prediction = recommend_crop(
        row["N"],
        row["P"],
        row["K"],
        row["temperature"],
        row["humidity"],
        row["ph"],
        row["rainfall"]
    )

    crop_predictions.append(prediction)

df["rule_crop"] = crop_predictions
```

```python
crop_means = df.groupby("label")[
    [
        "N",
        "P",
        "K",
        "temperature",
        "humidity",
        "ph",
        "rainfall"
    ]
].mean()


def data_driven_crop_recommendation(
    N, P, K, temperature, humidity, ph, rainfall
):

    input_values = np.array([
        N,
        P,
        K,
        temperature,
        humidity,
        ph,
        rainfall
    ])

    best_crop = None
    smallest_distance = float("inf")

    for crop in crop_means.index:

        crop_values = crop_means.loc[crop].values

        distance = np.sqrt(
            np.sum((input_values - crop_values) ** 2)
        )

        if distance < smallest_distance:
            smallest_distance = distance
            best_crop = crop

    return best_crop
```

```python
sample = df.iloc[0]

recommendation = data_driven_crop_recommendation(
    sample["N"],
    sample["P"],
    sample["K"],
    sample["temperature"],
    sample["humidity"],
    sample["ph"],
    sample["rainfall"]
)

print("Actual Crop:", sample["label"])
print("Recommended Crop:", recommendation)
```

```python
data_driven_predictions = []

for index, row in df.head(10).iterrows():

    prediction = data_driven_crop_recommendation(
        row["N"],
        row["P"],
        row["K"],
        row["temperature"],
        row["humidity"],
        row["ph"],
        row["rainfall"]
    )

    data_driven_predictions.append(prediction)

df.loc[df.index[:10], "data_driven_crop"] = data_driven_predictions

print(
    df[
        ["label", "data_driven_crop"]
    ].head(10).to_string(index=False)
)
```

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("Crop_recommendation.csv")

print("Dataset loaded successfully!")
print("Total records:", len(df))

# Selected features
features = [
    "N",
    "P",
    "K",
    "temperature",
    "humidity",
    "ph",
    "rainfall"
]

print("\nSelected Features:")
for feature in features:
    print(feature)


# Rule-based agricultural suitability function
def crop_decision(N, P, K, temperature, humidity, ph, rainfall):

    score = 0

    if N >= 40:
        score += 1
    elif N >= 20:
        score += 0.5

    if P >= 30:
        score += 1
    elif P >= 15:
        score += 0.5

    if K >= 30:
        score += 1
    elif K >= 15:
        score += 0.5

    if 20 <= temperature <= 30:
        score += 1
    elif 15 <= temperature <= 35:
        score += 0.5

    if 50 <= humidity <= 80:
        score += 1
    elif 40 <= humidity <= 90:
        score += 0.5

    if 5.5 <= ph <= 7.5:
        score += 1
    elif 5.0 <= ph <= 8.0:
        score += 0.5

    if 80 <= rainfall <= 250:
        score += 1
    elif 50 <= rainfall <= 300:
        score += 0.5

    if score >= 5:
        return "Highly Suitable"
    elif score >= 3:
        return "Moderately Suitable"
    else:
        return "Less Suitable"


# Process dataset records
decisions = []

for index, row in df.iterrows():

    decision = crop_decision(
        row["N"],
        row["P"],
        row["K"],
        row["temperature"],
        row["humidity"],
        row["ph"],
        row["rainfall"]
    )

    decisions.append(decision)

df["rule_decision"] = decisions


# Display first 10 results
print("\nFirst 10 Results:")
print(
    df[
        [
            "N",
            "P",
            "K",
            "temperature",
            "humidity",
            "ph",
            "rainfall",
            "label",
            "rule_decision"
        ]
    ].head(10).to_string(index=False)
)


# Count categories
print("\nDecision Category Counts:")
print(df["rule_decision"].value_counts())


# Dataset pattern analysis
print("\nAverage Conditions for Each Crop:")
print(
    df.groupby("label")[features]
    .mean()
    .round(2)
)
```
