```python
if nitrogen < 50:
    nitrogen_condition = "Low"
elif nitrogen < 100:
    nitrogen_condition = "Moderate"
else:
    nitrogen_condition = "High"

print("Nitrogen condition:", nitrogen_condition)
```

```python
if temperature < 20:
    temperature_condition = "Cold"
elif temperature <= 30:
    temperature_condition = "Moderate"
else:
    temperature_condition = "Hot"

print("Temperature condition:", temperature_condition)
```

```python
def classify_nitrogen(n):
    if n < 50:
        return "Low Nitrogen"
    elif n < 100:
        return "Moderate Nitrogen"
    else:
        return "High Nitrogen"


print(classify_nitrogen(30))
print(classify_nitrogen(75))
print(classify_nitrogen(120))
```

```python
def classify_environment(temp, humidity, rainfall):

    if temp < 20:
        temp_status = "Cold"
    elif temp <= 30:
        temp_status = "Moderate"
    else:
        temp_status = "Hot"

    if humidity < 40:
        humidity_status = "Low Humidity"
    elif humidity <= 80:
        humidity_status = "Moderate Humidity"
    else:
        humidity_status = "High Humidity"

    if rainfall < 100:
        rainfall_status = "Low Rainfall"
    elif rainfall <= 200:
        rainfall_status = "Moderate Rainfall"
    else:
        rainfall_status = "High Rainfall"

    return temp_status, humidity_status, rainfall_status


result = classify_environment(
    temperature,
    humidity,
    rainfall
)

print("Temperature:", result[0])
print("Humidity:", result[1])
print("Rainfall:", result[2])
```

```python
features = [
    "N",
    "P",
    "K",
    "temperature",
    "humidity",
    "ph",
    "rainfall"
]

# Example soil/environment conditions
input_values = np.array([
    90,      # Nitrogen
    42,      # Phosphorus
    43,      # Potassium
    25.5,    # Temperature
    80.2,    # Humidity
    6.5,     # pH
    200.5    # Rainfall
])

data_values = df[features].values

distances = np.sqrt(
    np.sum((data_values - input_values) ** 2, axis=1)
)

closest_index = np.argmin(distances)

recommended_crop = df.iloc[closest_index]["label"]

print("Recommended crop:", recommended_crop)
```

```python
features = [
    "N",
    "P",
    "K",
    "temperature",
    "humidity",
    "ph",
    "rainfall"
]

# Example soil/environment conditions
input_values = np.array([
    90,      # Nitrogen
    42,      # Phosphorus
    43,      # Potassium
    25.5,    # Temperature
    80.2,    # Humidity
    6.5,     # pH
    200.5    # Rainfall
])

data_values = df[features].values

distances = np.sqrt(
    np.sum((data_values - input_values) ** 2, axis=1)
)

closest_index = np.argmin(distances)

recommended_crop = df.iloc[closest_index]["label"]

print("Recommended crop:", recommended_crop)
```

```python
print("========== CROP RECOMMENDATION ==========")

print("\nInput Conditions:")

for feature, value in zip(features, input_values):
    print(feature, ":", value)

print("\nRecommended Crop:")
print(recommended_crop)

print("\nClosest Dataset Record:")
print(df.iloc[closest_index])
```
